# Hospital Patient Data Cleaning

```python
# loading the dataset into a dataframe

df_1 = pd.read_csv(r'\Users\oseni\Downloads\hospital_patient_data - hospital_patient_data_cleaning_practice.csv')
```

#### Discovery
##### There are 3000 rows and 8 columns, The data type of age is float instead of integer and it contains about 45 null cells
##### It was also discoved that city, and visit_count also contain some empty values and visit count has a string data type

```python
# check for duplicate data

duplicate_data = df_1.duplicated().sum()
print("Duplicate Data:", duplicate_data)
```

```python
# drop duplicate data

df_1 = df_1.drop_duplicates()
```

```python
# check if patient id has duplicates

df_1.duplicated(subset = ['Patient_ID']).sum()
```

```python
# observe columns using unique 

df_1['Gender'].unique()
```

```python
df_1['Gender'].value_counts()
```

```python
df_1['Record_Year'].unique()
```

```python
df_1['Visit_Type'].unique()
```

```python
df_1['City'].value_counts()
```

```python
df_1['City'].unique()
```

```python
# remove trailing spaces and change data format
df_1['City'] = df_1['City'].astype(str).str.strip().str.title()

# replace values with the correct format
df_1['City'] = df_1['City'].str.replace("Port-Harcourt", "Port Harcourt")

#remove wrong values
df_1 = df_1[df_1['City'] != 'Test']
```

```python
df_1['City'].unique()
```

```python
# fill up empty spaces
df_1['City'] = df_1['City'].str.strip().fillna("Unknown")
```

```python
df_1.info()
```

```python
# Clean age data

df_1['Age'] = pd.to_numeric(df_1['Age'], errors = 'coerce')
median_age = df_1['Age'].median()

df_1['Age'] = df_1['Age'].fillna(median_age).astype(int)

df_1.head(15)
```

```python
df_1.info()
```

```python
df_1['Visit_Count'].unique()
```

```python
# clean visit count and convert to proper format

df_1['Visit_Count'] = df_1['Visit_Count'].astype(str).str.replace('visits', '', case = False).str.strip()

word_to_num = {'six': '6',
              'seven': '7',
              'ten': '10',
              'nan': 'np.nan'}
df_1['Visit_Count'] = df_1['Visit_Count'].str.lower().replace(word_to_num)


df_1['Visit_Count'] = pd.to_numeric(df_1['Visit_Count'], errors = 'coerce')

df_1['Visit_Count'] = df_1['Visit_Count'].fillna(0).astype(int)

df_1.info()
```

```python
df_1.head(15)
```

```python
# create new column with a conditional based format

df_1['Patient_Category'] = np.where(df_1['Visit_Count'] >= 5, "Frequent Patient" , "Occasional Patient")
df_1.head(15)
```

```python
## Number of patients by City

df_1['City'].value_counts()
```

```python
## Number of Frequent Patients

frequent_patient = df_1['Patient_Category'].value_counts().max()
print("Number of Frequent Patient:", frequent_patient)
```

```python
## Average patient visit count

average_patient_visit = df_1['Visit_Count'].mean()
print("Average Visit Count:", average_patient_visit.round(3))
```

```python
## Number of Records remaining after cleaning

df_1.info()
```

```python
# Length of dataset after cleaning

Length = len(df_1)
print("Dataset after Cleaning:", Length)
```

```python
# create a summary table

number_of_patients = df_1['Patient_Category'].value_counts()
average_visit = df_1.groupby('Patient_Category')['Visit_Count'].mean()

pd.concat([number_of_patients,average_visit.round(3)], axis = 1, keys = ['Number of Patient', 'Average Visit'])
```

#### Observation
##### The table shows that the hospital has 1986 Frequent patient who visit on an average of 8.5 times, compared to 1005 Occasiona Patients who average 2.4 visits. 
##### This implies that the Frequent patient drive the vast majority of hospital daily traffic and workload despite making up to  two third of the headcounts. Consequently, hospitals need to create structures ans scheduling programmes to prevent overcrowding.

# Banking Customer Transaction Analysis

```python
# loading the dataset into a dataframe

df_2 = pd.read_csv(r'\Users\oseni\Downloads\Banking_Customer_Transation - Banking_Customer_Transation.csv')
```

```python
# data exploration

df_2.shape
```

```python
df_2.columns
```

```python
df_2.describe()
```

```python
df_2.info()
```

```python
# check for empty cells

df_2.isna().sum()
```

```python
# check for duplicates

duplicate_data = df_2.duplicated().sum()
print(duplicate_data)
```

```python
## Check if Customername are truly empty

df_2['Customer Name'] = df_2['Customer Name'].astype(str).str.strip()
df_2['Customer Name'] = df_2['Customer Name'].replace('', np.nan)
df_2['Customer Name'].isna().sum()
```

```python
# Check if city is truly empty

df_2['City'] = df_2['City'].astype(str).str.strip()
df_2['City'] = df_2['City'].replace('', np.nan)
df_2['City'].isna().sum()
```

```python
# Check if Transacction type is truly empty

df_2['Transaction Type'] = df_2['Transaction Type'].astype(str).str.strip().replace('', np.nan)
df_2['Transaction Type'].isna().sum()
```

```python
## Check if Account Status is truly empty

df_2['Account Status'] = df_2['Account Status'].astype(str).str.strip().replace('', np.nan)
df_2['Account Status'].isna().sum()
```

```python
# Check for spaces that are not empty and cant be coerced to numbers

df_2['Amount'] = df_2['Amount'].astype(str).str.strip().replace('', np.nan)

text_to_number = pd.to_numeric(df_2['Amount'], errors= 'coerce')

failed_row = df_2[df_2['Amount'].notna() & text_to_number.isna()]

len(failed_row)
```

```python
failed_row['Amount'].unique()
```

```python
# clean identified error

df_2['Amount'] = df_2['Amount'].str.replace('₦', '', regex = False)
df_2['Amount'] = df_2['Amount'].str.replace(',', '', regex = False)
df_2['Amount'] = df_2['Amount'].str.replace('O', '0', regex = False)

df_2['Amount'] = pd.to_numeric(df_2['Amount'], errors = 'coerce')
df_2['Amount'].isna().sum()
```

```python
# fill up empty cells

df_2['Customer Name'] = df_2['Customer Name'].fillna('Unknown')
df_2['City'] = df_2['City'].fillna('Unknown')
df_2['Transaction Type'] = df_2['Transaction Type'].fillna('Unknown')
df_2['Account Status'] = df_2['Account Status'].fillna('Unknown')
```

```python
df_2.info()
```

```python
# create new column using conditional based format

df_2['Transaction_Category'] = np.where(df_2['Amount'] >= 100000, "Large Transaction", "Regular Transaction")
df_2
```

```python
# Transactions in each city

df_2['City'].value_counts()
```

```python
# Average Transaction amount

transaction_amount = df_2['Amount'].mean()
print("Average Transaction Amount:", transaction_amount.round(3))
```

```python
# Transaction Category that are Large transactions

large_transactions = df_2['Transaction_Category'].value_counts()
large_transactions 
```

```python
# Value of Large transaction

print("Large Transaction:", large_transactions.min())
```

```python
# Transaction with the most frequen transaction type

frequent_transaction_type = df_2['Transaction Type'].value_counts().idxmax()
print("Most Frequent Transaction Type:", frequent_transaction_type)
```

```python
# Average transaction amount by transaction type

df_2.groupby('Transaction Type')['Amount'].mean().round(3)
```

```python
# Average transaction amount by City

df_2.groupby('City')['Amount'].mean().round(3)
```

```python
# city with the highest average transaction amount

highest_average_city = df_2.groupby('City')['Amount'].mean().round(3)
print("City with the highest average transaction amount:", highest_average_city.idxmax())
```

##### Findings
##### This showed that Benin had the highest average transaction amount meaning that, they spend more money per transaction than any other city making it the most valuable region to generate high revenue. 
##### The business could decide to allocate  budget to improve the inventory level to maximize returns

```python
# removing duplicates

df_3 = df_3.drop_duplicates()
```

```python
df_3.info()
```

```python
# check for hidden texts or values

df_3['City'].unique()
```

```python
# check if the empty cells are actually empty

df_3['City'] = df_3['City'].astype(str).str.strip()
df_3['City'] = df_3['City'].replace('', np.nan)
df_3['City'].isna().sum()
```

```python
df_3.info()
```

```python
# check if customer type is actually empty

df_3['Customer Type'] = df_3['Customer Type'].astype(str).str.strip()
df_3['Customer Type'] = df_3['Customer Type'].replace('', np.nan)
df_3['Customer Type'].isna().sum()
```

```python
# check for rows that failed to format to numbers

df_3['Monthly Consumption'] = df_3['Monthly Consumption'].astype(str).str.strip().replace('', np.nan)

text_to_number = pd.to_numeric(df_3['Monthly Consumption'], errors= 'coerce')

failed_row = df_3[df_3['Monthly Consumption'].notna() & text_to_number.isna()]

len(failed_row)
```

```python
failed_row['Monthly Consumption'].unique()
```

```python
# Cleaning the textual symbols

df_3['Monthly Consumption'] = df_3['Monthly Consumption'].str.replace(',', '', regex = False)
df_3['Monthly Consumption'] = df_3['Monthly Consumption'].str.replace('O', '0', regex = False)

df_3['Monthly Consumption'] = pd.to_numeric(df_3['Monthly Consumption'], errors = 'coerce')
df_3['Monthly Consumption'].isna().sum()
```

```python
# check if Account Status is actually empty

df_3['Account Status'] = df_3['Account Status'].astype(str).str.strip()
df_3['Account Status'] = df_3['Account Status'].replace('', np.nan)
df_3['Account Status'].isna().sum()
```

```python
# check if Energy Source is actually empty

df_3['Energy Source'] = df_3['Energy Source'].astype(str).str.strip()
df_3['Energy Source'] = df_3['Energy Source'].replace('', np.nan)
df_3['Energy Source'].isna().sum()
```

```python
# Check if the are all empty at some point

check_blank_rows = ['City',	'Customer Type', 'Monthly Consumption',	'Account Status', 'Energy Source']
df_3 = df_3.dropna(subset = check_blank_rows, how ='all')
len(df_3)
```

```python
# Fill up dataset

df_3['City'] = df_3['City'].fillna('Not Provided')
df_3['Customer Type'] = df_3['Customer Type'].fillna('Unknown')
df_3['Account Status'] = df_3['Account Status'].fillna('Unknown')
df_3['Energy Source'] = df_3['Energy Source'].fillna('Not provided')
```

```python
df_3.info()
```

```python
# Create consumption category

df_3['Consumption_Category'] = np.where(df_3['Monthly Consumption'] >= 500, "High Consumption", "Low Consumption")
df_3
```

```python
# Number of customers in each city.

df_3['City'].value_counts()
```

```python
# Average consumption.

average_consumption = df_3['Monthly Consumption'].mean()
print("Average Consumption:", average_consumption.round(3) )

```

```python
# Number of high-consumption customers.

high_consumption_users = df_3['Consumption_Category'].value_counts().min()
print("Number of High consumption Users:", high_consumption_users )
```

```python
# Average consumption by customer type.

average_consumption_customertype = df_3.groupby('Customer Type')['Monthly Consumption'].mean()
print(average_consumption_customertype.round(3))
```

```python
# Number of customers using each energy source

df_3['Energy Source'].value_counts()
```

```python
# creating summary table

customer_type = df_3['Customer Type'].value_counts()
average_Consumption = df_3.groupby('Customer Type')['Monthly Consumption'].mean()

pd.concat([customer_type, average_Consumption.round(3)], axis = 1, keys = ['Number of Customers', 'Average Consumption'])
```

##### Findings

##### Industrial customers represernts the smallest group yet drive the highest average consumption by 6464.326kwh per month
##### Residential users make up the vast majority of customer base yet have the lowest individual elctricity consumption footprint
##### The electricity company should redefine industrial operational factors to redefine high demand grid forecasts
